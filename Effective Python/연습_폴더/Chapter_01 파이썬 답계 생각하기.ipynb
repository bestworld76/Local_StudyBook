##### Item 03 파이썬이 컴파일 시점에 오류를 감지하리라고 절대로 기대하지 말라

```python
# Error 발생하는 경우
def bad_reference():
    print(my_var)
    my_var = 123

bad_reference()
```

```python
# 정상적인 경우
def sometimes_ok(s):
    if s:
        my_var = 123
    print(my_var)

sometimes_ok(True)
```

#### Item 04 복잡한 식을 쓰는 대신 도우미 함수를 작성하라

```python
from urllib.parse import parse_qs

my_values = parse_qs(
    "적=5&청=0&녹=",
    keep_blank_values=True)

print(repr(my_values))
```

```python
print(f'적    : {my_values.get('적')}')
print(f'녹    : {my_values.get('녹')}')
print(f'투명도 : {my_values.get('투명도')}')
```

```python
red_str = my_values.get('적',[""])
red = int(red_str[0]) if red_str[0] else 0

print(red)
```
