##### Item 17 range 보다는 enumerate를 사용하라

```python
from random import randint

random_bits = 0
for i in range(32):
    if randint(0, 1):
        random_bits |= 1 << i

print(bin(random_bits))
```

##### Item 18 여러 이터레이터를 대해 나란히 루프를 수행하려면 zip을 사용하라

```python
names = ['Cecilia', '남궁민수', '帽宅仝']
counts = [len(n) for n in names]
print(counts)
```

```python
longest_name = None
max_count = 0

for i in range(len(names)):
    count = counts[i]
    if count > max_count:
        longest_name = names[i]
        max_count = count

print(longest_name)
```

```python
for i , name in enumerate(names):
    count = counts[i]
    if count > max_count:
        longest_name = name
        max_count = count
print(longest_name)
```

```python
longest_name = None
max_count = 0

for name, count in zip(names, counts):
    if count > max_count:
        longest_name = name
        max_count = count

print(longest_name)
```

```python
names.append('Rosalind')
for name in names:
    if name == 'Rosalind':
        names.remove(name)
    print(name)
```

##### Item 21 인자에 대해 이터레이션살 때는 방어적으로 접근하라

```python
def normalize(number):
    total = sum(number)
    result = []
    for value in number:
        percent = 100 * value / total
        result.append(percent)
    return result

visits =  [15,35,80]
percentages = normalize(visits)
print(percentages)
assert sum(percentages) == 100.0
```

```python
def read_visits(data_path):
    with open(data_path) as f:
        for line in f:
            yield int(line)

it = read_visits("my_number.txt")
percentages = normalize(it)
print(percentages)
```

##### Item 23 효율적인 쇼트 서트 로직을 사용하려면 any나 all에게 이터레이터를 전달하라

```python
import random

def filip_coin():
    if random.randint(0,1) == 0:
        return 'Head'
    else:
        return 'Tail'

def flip_is_heads():
    return filip_coin() == 'Head'

all_heads = True

for _ in range(100):
    if not flip_is_heads():
        all_heads = False
        break

print(f"모두 참일 때 : {all([1,2,3])}")
```

```python
all_heads = all(flip_is_heads() for _ in range(20))
```

```python
def repeater_is_head(count):
    for _ in range(count):
        yield flip_is_heads()

all_heads = all(repeater_is_head(20))
```

##### Item 24 이터레이터나 제너레이터를 다룰 때는 itertools 를 사용하라

```python
import itertools

it = itertools.chain([1,2,4], [4,5,6])
print(list(it))
```

```python
it1 = [i * 3 for i in ('a', 'b', 'c')]
it2 = [j * 2 for j in ('x', 'y', 'z')]
nested_it = [it1, it2]
print(nested_it)
output_it = itertools.chain.from_iterable(nested_it)
print(output_it)
print(list(output_it))
```

```python
it = itertools.repeat('hello', 3)
print(list(it))
```

```python
it = itertools.cycle([1,2])
result = p=[next(it) for _ in range(9)]
print(result)
```

```python
it1, it2, it3 = itertools.tee(['first', 'second'], 3)
print(list(it1))
print(list(it2))
print(list(it3))
```

```python
keys = ['one', 'two', 'three']
values = [1, 2]

normal = list(zip(keys, values))
print("zip:        ", normal)

it = itertools.zip_longest(keys, values, fillvalue="nope")
longest = list(it)
print("zip_longest:", longest)
```

```python
values = [x for x in range(1,11)]
first_five = itertools.islice(values, 5)
print(f"맨 앞 5개 : {list(first_five)}")
print('-----------------------------')
middle_odds = itertools.islice(values, 2, 8, 2)
print(f"중간 홀 수들 : {list(middle_odds)}")
```

```python
less_than_seven = lambda x : x < 7
it = itertools.takewhile(less_than_seven, values)
print(list(it))
```

```python
it = itertools.batched(values, 3)
print(list(it))
```

```python
route = ['seoul', 'daegu', 'daejeon', 'busan']
it = itertools.pairwise(route)
print(list(it))
```

```python
it = itertools.permutations(route, 2)
print(list(it))
```
