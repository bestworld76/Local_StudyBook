```python
votes = {
    "수달"  : 1281,
    "북극곱" : 587,
    "여우"   : 863,
}

def populate_rank(votes, rank):
    names = list(votes.keys())
    names.sort(key=votes.get, reverse=True)
    for i, name in enumerate(names, 1):
        rank[name] = i

def get_winner(rank):
    return next(iter(rank))

ranks = {}
populate_rank(votes, ranks)
print(ranks)
winner = get_winner(ranks)
print(winner)
```

```python
from _collections_abc import MutableMapping

class SortedDict(MutableMapping):
    def __init__(self):
        self.data = {}

    def __getitem__(self, key):
        return self.data[key]

    def __setitem__(self, key, value):
        self.data[key] = value

    def __delitem__(self, key):
        del self.data[key]

    def __iter__(self):
        keys = list(self.data.keys())
        keys.sort()
        for key in keys:
            yield key

    def __len__(self):
        return len(self.data)

sorted_ranks = SortedDict()
populate_rank(votes, sorted_ranks)
print(sorted_ranks)
print(sorted_ranks.data)
winner = get_winner(sorted_ranks)
print(winner)
```

```python
counters = {
    '펌퍼니클' : 2 ,
    '샤워도'   : 1,
}

key =  '밀'

if key in counters:
    count = counters[key]
else:
    count = 0

counters[key] = count + 1
print(counters)
```

##### Item 27 내부 상태에서 원소가 없는 경우를 처리할 때는 setdefault 보다 defaultdict 를 사용해라

```python
visits = {
    "Korea" : {"Daejeon": "Andong"},
    "Japan" : {"Hakone"}
}
```

```python
# 짧다
visits.setdefault('Prance', set()).add('아를')
print(visits)

# 길다
if (japan := visits.get('Japna')) is None:
    visits['Japan'] = japan = set()

japan.add('교토')
print(visits)
```

```python
class Visits:
    def __init__(self):
        self.data = {}

    def add(self, country, city):
        city_set = self.data.setdefault(country, set())
        city_set.add(city)
```

```python
visits = Visits()
visits.add('러시아', '예카텐린부르크')
visits.add('탄자니아', '잔지바르')
print(visits.data)
```

```python
from collections import defaultdict

class Visits:
    def __init__(self):
        self.data = defaultdict(set)

    def add(self, country, city):
        self.data[country].add(city)
```

```python
visits = Visits()
visits.add("영국", "베스")
visits.add("영국", "런던")
print(visits.data)
```

##### Item 29 딕셔너리, 리스트, 튜플을 깊게 중첩 시키기보다는 클래스를 합성하라

```python
class SimpleGradebook:
    def __init__(self):
        self._grades = {}

    def add_student(self, name):
        self._grades[name] = []

    def report_grade(self, name, score):
        self._grades[name].append(score)

    def average_grade(self, name):
        grade = self._grades[name]
        return sum(grade) / len(grade)

book = SimpleGradebook()
book.add_student("아이작 뉴튼")
book.report_grade("아이작 뉴튼", 90)
book.report_grade("아이작 뉴튼", 95)
book.report_grade("아이작 뉴튼", 85)

print(book.average_grade("아이작 뉴튼"))
```

```python
class BySubjectGradebook: 
    def __init__(self):
        self._grades = {}
        
    def add_student(self, names):
        self._grades[names] = defaultdict(list) # 내부 dict 
        
    def report_grade(self, name, subject, grade):
        by_subject = self._grades[name]
        grade_list = by_subject[subject]
        grade_list.append(grade)
        
    def average_grade(self, name):
        by_subject = self._grades[name]
        total, count = 0, 0
        for grades in by_subject.values():
            total += sum(grades)
            count += len(grades)
        return total / count
```
