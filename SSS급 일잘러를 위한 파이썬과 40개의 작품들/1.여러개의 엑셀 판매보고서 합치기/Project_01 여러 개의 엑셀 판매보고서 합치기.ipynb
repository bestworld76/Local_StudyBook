### 여러 개의 엑셀 판매보고서 합치기

#### 라이브러리 임포트하기

```python
import glob
import os
from openpyxl import load_workbook
from openpyxl import Workbook
```

#### 파일 경로 설정하기

```python
folder_path = '/Users/bestw/Desktop/ex/python_ex/PythonProject/개인학습/git_local_folder/SSS급 일잘러를 위한 파이썬과 40개의 작품들/1.여러개의 엑셀 판매보고서 합치기'

file_list = [file for file in os.listdir(folder_path) if file.endswith('.xlsx') and not file.startswith('result_file') and not file.startswith('결과.xlsx')]
print(folder_path, file_list )
```

```python
result_excel_file = glob.glob(f'{folder_path}/result_file.xlsx')

market_list      = []
date_list        = []
total_salary_sum = []

for file in file_list:
    wb = load_workbook(file, data_only=True)
    ws = wb.active
    market_list.append(ws['B1'].value)
    date_list.append(ws['B2'].value)
    total_salary_sum.append(ws['B3'].value)

print(market_list)
print(date_list)
print(total_salary_sum)

try:
    wb = load_workbook(f'{folder_path}/result_file.xlsx', data_only=True)
    ws = wb.active
except:
    wb = Workbook()
    ws = wb.active

for i in range(len(market_list)):
 ws.cell(row = i + 1, column=1).value = market_list[i]
 ws.cell(row = i + 1, column=2).value = date_list[i]
 ws.cell(row = i + 1, column=3).value = total_salary_sum[i]

wb.save(f'{folder_path}/result_file.xlsx')
```
