## 8.1 기본 파이썬 기능으로 테이블 저장하기

```python
# code 8-1 파이썬 딕셔너리 자료 구조로 테이블 저장하기
fish_measures = {
    'Fish'   : ['Angelfish', 'Zebrafish', 'Killfish', 'Swordfish'],
    'Length' : [15.2, 6.5, 9, 6],
    'Width'  : [7.7, 2.1, 4.5, 2]
}

# code 8-2 딕서너리로 테이블 접급하기
zebrafish_index = fish_measures['Fish'].index('Zebrafish')
zebrafish_length = fish_measures['Length'][zebrafish_index]
print(f' {zebrafish_index} : zebrafish 길이는 {zebrafish_length:.2f} cm')
```

## 8.2 판다스로 테이블 탐색하기

```python
# code 8-3 판다스 라이브러리 블러오기
import pandas as pd
import numpy as np

# code 8-4 테이블을 판다스의 데이터 프레임 형식으로 변환하기
df = pd.DataFrame(fish_measures)
print(df)
```

```python
# code 8-6 숫자열 요약하기
df.describe()
```

```python
# code 8-7 열의 평균 계산하기
print(df.iloc[:,1:].mean())
```

```python
# code 8-8 문자열 요약하기
print(df.describe(include=[np.object_]))
```

```python
# code 8-9 테이블을 2차원 넘파이 배열로 가져오기
print(df.values)
assert type(df.values) == np.ndarray
```

## 8.5 테이블의 행과 열 수정하기

```python
df.set_index('Fish', inplace=True)
```

```python
# code 8-22 행과 열 바꾸기
df_transposed =df.T
print(df_transposed)
```

```python
# code 8-24 신규 열 추가하기
df_transposed['Clownfish'] = [10.6, 3.7]
print(df_transposed)
```

```python
# code 8-25 신규 열을 여러 개 추가하기
df_new = df_transposed.assign(Clownfish2=[10.6,3.7], Clownfish3=[10.6,3.7])
assert 'Clownfish2' not in df_transposed.columns
assert 'Clownfish2' in df_new.columns
print(df_new)
```

```python
# code 8-26 여러 열 삭제하기
df_new.drop(columns=['Clownfish2','Clownfish3'], inplace=True)
df_new
```
