# 5장 사이파이를 사용한 기본 확률 및 통계분석

#### Code 5-1 사이파이의 stats 모듈 불러오기

```python
from scipy import stats
from scipy.stats import binom
```

#### Code 5-2 사이파이로 극단적인 동전 앞면 개수 분석하기

```python
num_heads = 16
num_flips = 20
prob_head = 0.5
prob = stats.binomtest(num_heads, num_flips, prob_head)
print(f"15개 이상의 동전 앞면 또는 뒷면이 관찰될 확률은 {prob.pvalue:.17f}입니다.")
```

#### Code 5-3 stats.binom.pmf 정확한 확률 계산하기

```python
prob_16_heads = stats.binom.pmf(num_heads, num_flips, prob_head)
print(f"{num_heads}번의 앞면 중 {num_flips}번이 관측될 확율은 {prob_16_heads} 입니다")
```
