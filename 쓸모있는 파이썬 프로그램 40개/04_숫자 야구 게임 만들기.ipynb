# 제시한 숫자에 대해서 스크라이크, 볼 판정하기

```python
import random

# 랜덤 숫자 3개 생성 (중복 없이)
numbers = list(range(0, 10))
random.shuffle(numbers)
three_numbers = random.sample(numbers, 3)
# print(f"three_numbers: {three_numbers}")  # 디버깅용, 실제 게임할 땐 주석 처리

round = 1

while True:
    strike = 0
    ball = 0

    print(f"\n[ {round} 라운드 ]\n")

    # 사용자 입력 처리 (3,4,5 또는 3 4 5 또는 345 등)
    user_input = input("0~9 사이 숫자 3개 입력 (예: 3,4,5 또는 345 등) >> ")
    user_input = user_input.replace(',', '').replace(' ', '')

    # 숫자 3개 제대로 입력했는지 확인
    if len(user_input) != 3 or not user_input.isdigit():
        print("❗ 숫자 3개를 정확히 입력해주세요.\n")
        continue

    num1, num2, num3 = map(int, user_input)
    list1 = [num1, num2, num3]

    # 중복 체크
    if len(set(list1)) != 3:
        print("❗ 중복 없이 3개의 숫자를 입력해주세요.\n")
        continue

    # 스트라이크 & 볼 계산
    for i in range(3):
        if three_numbers[i] == list1[i]:
            strike += 1
        elif list1[i] in three_numbers:
            ball += 1

    print(f"🎯 Strike: {strike}, Ball: {ball}\n")

    if strike == 3:
        print(f"[Game End] {round} 라운드만에 맞췄습니다! 🎉")
        break

    round += 1
```
