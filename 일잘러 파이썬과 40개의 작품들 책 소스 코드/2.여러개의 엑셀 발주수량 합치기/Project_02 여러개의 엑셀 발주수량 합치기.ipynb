```python
import pandas as pd
import os
import glob
from openpyxl import load_workbook

path = os.chdir('/Users/bestw/Desktop/ex/python_ex/PythonProject/개인학습/git_local_folder/일잘러 파이썬과 40개의 작품들 책 소스 코드/2.여러개의 엑셀 발주수량 합치기')
order_files = [file for file in os.listdir(path)
               if file.endswith('.xlsx') and
               not file.startswith('통합') and
               not file.startswith('result')]

print(f"order_files: {order_files}")

order_position_list = []
order_product_list = []
order_quantity_list = []

for file in order_files:
    wb = load_workbook(file, data_only=True)
    ws = wb.active

    order_position = ws['B1'].value
    print(f'order_position : {order_position}')

    for order_product, order_quantity in ws.iter_rows(
        min_row=4,
        max_row=12,
        min_col=1,
        max_col=2,
        values_only=True
    ):
        if order_product is not None:
            order_position_list.append(order_position)
            order_product_list.append(order_product)
            order_quantity_list.append(order_quantity)

    wb.close()

print(f'order_position_list : {order_position_list}')
print(f'order_product_list : {order_product_list}')
print(f'order_quantity_list : {order_quantity_list}')

df = pd.DataFrame({
    'order_position': order_position_list,
    'order_product': order_product_list,
    'order_quantity': order_quantity_list
})

df.head()
```
