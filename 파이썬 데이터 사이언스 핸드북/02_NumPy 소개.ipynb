```python
import pandas as pd

def read_csv_file(filename):
    return pd.read_csv(f"{filename}.csv", encoding='utf-8', low_memory=False)

df = read_csv_file('president_heights')
df.head()
```

```python
heights = np.array(df['height(cm)'])
print(f'Mean height: {heights.mean()}')
print(f'Standard deviation: {heights.std()}')
print(f'Minimum height: {heights.min()}')
print(f'Maximum height: {heights.max()}')
```

```python
print(f'25th percentile : {np.percentile(heights, 25)}')
print(f'Median : {np.median(heights)}')
print(f'75th percentile : {np.percentile(heights, 75)}')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns; sns.set()

plt.hist(heights)
plt.title('Heights Distribution of US President Heights')
plt.xlabel('Heights (CM)')
plt.ylabel('Number')
plt.show()
```
