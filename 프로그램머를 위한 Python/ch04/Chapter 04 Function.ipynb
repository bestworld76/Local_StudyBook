## 4.5 사례 연구 : 확률 게임

```python
"""크랩스 주사위 게임 시뮬레이션하기"""

def roll_dice():
    """ 두 개의 주사위를 던지고 튜플로 각 주사위의 숫자를 반환한다. """
    die1 = random.randrange(1,7)
    die2 = random.randrange(1,7)
    return (die1, die2)

def display_dice(dice):
    """ 두 개의 주사위 숫자를 한번에 출력한다. """
    die1, die2 = dice # 튜 플을 변수 die1과 die2에 푼다
    print(f"플레어가 던진 주사위는 {die1} + {die2} = {sum(dice)}")

die_values = roll_dice() # 첫 번째 게임
display_dice(die_values)

# 첫 번째 게임을 보고 게임 승패 또는 포인트 점수를 결정한다.
sum_of_dice = sum(die_values)

if sum_of_dice in (7, 11): #승리
    game_state = '승리'
elif sum_of_dice in (2,3,12): #패배
    game_state = '패배'
else: # 포인트 값 저장
    game_state = "계속"
    my_point = sum_of_dice
    print(f"나의 포인트는 : {my_point} 입니다")

# 플레이어가 이기거나 질 때까지 반복한다.
while game_state == '계속':
    die_values = roll_dice()
    display_dice(die_values)
    sum_of_dice = sum(die_values)

    if sum_of_dice == my_point: # 포인트 값을 만들면 승리한다.
        game_state = '승리'
    elif sum_of_dice == 7: game_state = '패배'

# 승패에 대한 메시지 출력한다.
if game_state == '승리':
    print('플레이어 승리')
else:
    print('플레이어 패배')

```
